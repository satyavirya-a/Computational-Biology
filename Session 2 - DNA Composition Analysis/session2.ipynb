## GC & AT Composition
[]()
Komposisi ini memberitahu sifat tahan panas:

- GC rich content -> Lebih tahan suhu panas, umumnya DNA lebih stabil (Hidrogen Bond rangkap 3)
- AT rich content -> kurang tahan suhu panas, umumnya DNA kurang stabil (Hidrogen Bond rangkap 2)

Faktor stabil: 
- GC & AT Comp
- Salt concentration
- Base Stacking
- Suhu

```python
from Bio.Seq import Seq
from Bio.SeqUtils import GC
```

```python
seqA = "GCCGATCC" #len = 8
```

```python
#menghitung manual persentase basa nitrogen GC
manual = (2+4) / 8 * 100
print(f'GC Composition = {manual}%')
```

```python
#menghitung pakai formula persentase basa nitrogen GC
formula = (seqA.count("G") + seqA.count("C")) / len(seqA) * 100
print(f'GC Composition = {formula}%')
```

```python
#pakai function yang di import
print(f'GC Composition = {GC(seqA)}%')
```

```python
# untuk AT pakai komplemen nya aja, bisa manual juga atau engga
print(f'GC Composition = {100 - GC(seqA)}%')
```

## Melting Temperature
[Sumber Dokumetasi Library Melting Temperature](https://biopython.org/docs/latest/api/Bio.SeqUtils.MeltingTemp.html#Bio.SeqUtils.MeltingTemp.Tm_Wallace)

ada proses denaturation (terbuka DNA awalnya double helix -> single strand)\
Melting temp -> kondisi 50% dari dna kondisi terbuka\
Tujuan terbuka -> replikasi DNA

Di tubuh, ada enzim PCR (Polymerase Chain Reaction) membantu membuka DNA. Kalau di lab, pengaruh suhu yang diadjust



### Metode Wallace

Tm Wallace = 2 * (jumlah A + jumlah T) + 4 * (jumlah G + jumlah C)

```python
#definisi manual
def tm_wallace(seq):
    count_A = seq.count("A") 
    count_T = seq.count("T")
    count_G = seq.count("G")
    count_C = seq.count("C")
    
    return (2*(count_A + count_T) + 4 * (count_G + count_C))
```

```python
print(f"TM Wallace dari seqA = {tm_wallace(seqA)} ℃")
```

#### Coba coba metode wallace

```python
seqA = "GCCGATCG" # GC Rich
seqB = "GCGCATAT" # balance
seqC = "ATTACGAT" # AT Rich

def print_tm(seqA, seqB, seqC, method):
    print(f"TM {method} seqA (GC Rich): {seqA} ℃")
    print(f"TM {method} seqB (Balance): {seqB} ℃")
    print(f"TM {method} seqC (AT Rich): {seqC} ℃")

print_tm(tm_wallace(seqA), tm_wallace(seqB), tm_wallace(seqC), "wallace")
```

yang gc rich lebih stabil dari yang lain

```python
#bisa juga pake function
from Bio.SeqUtils import MeltingTemp as mt
```

```python
print_tm(mt.Tm_Wallace(seqA), mt.Tm_Wallace(seqB), mt.Tm_Wallace(seqC), "wallace")
```

### Metode GC-based

Dilihat dari komposisi GC dan garam (NA), kalau dikasih NA bisa mengurangi phospor repultion

```python
print_tm(mt.Tm_GC(seqA), mt.Tm_GC(seqB), mt.Tm_GC(seqC), "GC-based")
```

### Metode Nearest Neighbor (NN)
based on thermodynamic (entalpy & entropy). Ada faktor base stacking (tumpukan basa nitrogen)

```python
print_tm(mt.Tm_NN(seqA), mt.Tm_NN(seqB), mt.Tm_NN(seqC), "Nearest Neighbor")
```

## Molecular Weight
[Sumber Dokumentasi Library Molecular Weight](https://biopython.org/docs/latest/api/Bio.SeqUtils.html#Bio.SeqUtils.molecular_weight)

menghitung massa relatif dari molekur nucleotide (satuan Da (Dalton))

`formula = (jumlah A * massa A) + (jumlah T * massa T) + (jumlah G * massa G) + (jumlah C * massa C)`

Tapi, massanya tidak murni. Ketika terbentuk DNA, nukleotida bertemu sesamanya dan melepaskan gugus kimia hasil reaksi. Jadi, massa berkurang jadi massa residu

```python
#pakai function aja
from Bio.SeqUtils import molecular_weight
```

```python
print(molecular_weight(seqA), " Da") # untuk single strand, bukan massa double helix

print(molecular_weight(seqA, double_stranded=True), " Da") # untuk double strand
                                                            # bukan 2 kali karena pasangan komplemeter jadi beda

print(seqA)
```
