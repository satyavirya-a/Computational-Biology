## Sequence Manipulation

```python
seqA = Seq("CGATTGACT")
```

```python
# Cari tahu index basa nitrogen pertama kali ketemu
idxA  = seqA.index("G")
idxA
```

```python
# Cari jumlah suatu basa nitrogen
countA = seqA.count("T")
countA
```

## Nucleotide Frequency Plot

Mengapa perlu mencari frequency ? Ada masing masing karakteristik \
\
CG banyak -> Heat resistence, Susah dibuka \
AT lebih banyak ->  mudah dibuka, tidak tahan panas, tidak stabil

```python
import matplotlib.pyplot as plt
from collections import Counter
# -> Counter setiap item di Seq

freq_dna = Counter(seqA)

print(freq_dna) #tipenya dictionary
```

Visualisasi Hasil Perhitungan Menggunakan Matplotlib

```python
plt.bar(freq_dna.keys(), freq_dna.values())
plt.show()
```
